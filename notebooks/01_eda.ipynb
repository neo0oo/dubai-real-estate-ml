Exploratory Data Analysis 

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("../data/processed/cleaned_dubai_properties.csv")
df.head()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df['rent'], bins=50, color='blue', edgecolor='black')
plt.xlabel('Rent(AED)')
plt.ylabel('Frequency')
plt.title('Distribution of Rent of Dubai Properties')
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.hist(np.log1p(df['rent']), bins=50, color='blue',edgecolor='black')
plt.xlabel('Log(Rent)')
plt.ylabel('Frequency')
plt.title('Distribution of Log(Rent)')
plt.savefig('../reports/figures/Distribution_of_Rent.png',dpi=300,bbox_inches='tight')

plt.show()
```

```python
df.describe(include='all')
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True),annot=True,fmt=".2f",cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['area_in_sqft'], df['rent'], alpha=0.3)
plt.xlabel('Area (sq ft)')
plt.ylabel('Rent (AED)')
plt.title('Area vs Rent')

plt.savefig('../reports/figures/area_vs_rent.png',dpi=300,bbox_inches='tight')

plt.show()
```
